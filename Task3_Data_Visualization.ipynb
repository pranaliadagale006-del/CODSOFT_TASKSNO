```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    'Student': ['Rahul', 'Priya', 'Amit', 'Sneha', 'Neha'],
    'Marks': [85, 90, 75, 95, 80]
}

df = pd.DataFrame(data)

print(df)

plt.figure(figsize=(8, 5))
plt.bar(df['Student'], df['Marks'])
plt.title('Student Marks Comparison')
plt.xlabel('Student Name')
plt.ylabel('Marks')
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.plot(df['Student'], df['Marks'], marker='o')

plt.title('Student Marks Line Chart')
plt.xlabel('Student Name')
plt.ylabel('Marks')

plt.grid()
plt.show()
```

```python
plt.figure(figsize=(7, 7))

plt.pie(
    df['Marks'],
    labels=df['Student'],
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Student Marks Distribution')
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df['Marks'], bins=5, edgecolor='black')

plt.title('Distribution of Student Marks')
plt.xlabel('Marks')
plt.ylabel('Number of Students')

plt.show()
```

```python
import matplotlib.pyplot as plt

students = ['Rahul', 'Priya', 'Amit', 'Sneha', 'Neha']
marks = [85, 90, 75, 95, 80]

plt.figure(figsize=(8, 5))

plt.scatter(students, marks, color='red', marker='o', s=100)

plt.title('Student Marks Scatter Plot')
plt.xlabel('Students')
plt.ylabel('Marks')

plt.grid(True)
plt.show()
```

```python
Insights from Charts

1. Bar Chart
The bar chart shows the comparison between different categories.
Some categories have higher values than others.

2. Line Chart
The line chart shows the changes in data over time.
It helps to identify trends and patterns.

3. Pie Chart
The pie chart shows the percentage distribution of different categories.
It helps to understand the contribution of each category.

4. Histogram
The histogram shows the frequency distribution of data.
It helps to understand how values are distributed.

5. Scatter Plot
The scatter plot shows the relationship between two variables.
It helps to identify patterns and correlations between data points.

    Final Conclusion
The data visualization helps to understand data easily. 
Different charts provide useful insights, identify patterns, compare values, and understand relationships between variables.
```
