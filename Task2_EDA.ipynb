```python
import pandas as pd
print("Pandas Imported Successfully")
```

```python
data = {'Name':['rahul','Priya','Amit','Sneha','Neha'],
        'Marks':[85,90,75,95,80],
        'Age':[20,21,20,22,21]
       }
df = pd.DataFrame(data)
print(df)
```

```python
print("Mean:",df['Marks'].mean())
print("Median:",df['Marks'].median())
print("Mode:",df['Marks'].mode()[0])
```

```python
print("Highest Marks:",df['Marks'].max())
print("Lowest Marks:",df['Marks'].min())
```

```python
print(df.describe())
```

```python
import matplotlib.pyplot as plt
plt.bar(df['Name'],df['Marks'])
plt.title('Student Marks Analysis')
plt.xlabel('Student Name')
plt.ylabel('Marks')
plt.show()
```

```python
plt.hist(df['Marks'],bins=5)
plt.title('Distribution of Student Marks')
plt.xlabel('Marks')
plt.ylabel('Number of Students')
plt.show()
```

```python
print("EDA Findings")
print("1. Total Students:", len(df))
print("2. Average Marks:", df['Marks'].mean())
print("3. Highest Marks:", df['Marks'].max())
print("4. Lowest Marks:", df['Marks'].min())
print("5. Student with Highest Marks:", df.loc[df['Marks'].idxmax(), 'Name'])
```
